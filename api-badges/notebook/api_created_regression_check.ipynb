# API-created reproducible regression check

This notebook uses only Python's standard library to generate deterministic synthetic data, fit a simple linear relationship, and verify the result.

```python
import csv
from pathlib import Path

points = [(x, 3.5 + 2.25 * x) for x in range(1, 31)]
x_mean = sum(x for x, _ in points) / len(points)
y_mean = sum(y for _, y in points) / len(points)
slope = sum((x - x_mean) * (y - y_mean) for x, y in points) / sum((x - x_mean) ** 2 for x, _ in points)
intercept = y_mean - slope * x_mean
assert abs(slope - 2.25) < 1e-12
assert abs(intercept - 3.5) < 1e-12
print({'intercept': intercept, 'slope': slope, 'rows': len(points)})
```

```python
output = Path('/kaggle/working/api_regression_predictions.csv')
with output.open('w', newline='', encoding='utf-8') as handle:
    writer = csv.writer(handle)
    writer.writerow(['x', 'target', 'prediction'])
    for x, target in points:
        writer.writerow([x, target, intercept + slope * x])
print(f'Wrote {output}')
```
